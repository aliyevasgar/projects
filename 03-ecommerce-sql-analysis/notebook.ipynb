```python
import os
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
fig_size = (12, 6)
os.makedirs("images", exist_ok=True)
```

```python
top_products = pd.read_csv("best_selling_product")

plt.figure(figsize=fig_size)
ax = sns.barplot(
    data=top_products,
    x="total_sales",
    y="name",
    hue="name",
    palette="crest",
    legend=False,
)

min_val = top_products["total_sales"].min() - 10
max_val = top_products["total_sales"].max() + 5
plt.xlim(min_val, max_val)
plt.xticks(np.arange(min_val, max_val + 1, step=5))

ax.xaxis.grid(True, linestyle="--", alpha=0.4, color="#90a4ae")
ax.set_axisbelow(True)

plt.title("Top 10 Products by Sales", fontsize=14, color="#263238", pad=15)
plt.xlabel("Total Sales", color="#37474f")
plt.ylabel("")
plt.tight_layout()
plt.savefig("images/top_products.png", dpi=150, bbox_inches="tight")
plt.show()
```

```python
monthly_revenue = pd.read_csv("monthly_trend")
monthly_revenue["monthly_trend"] = pd.to_datetime(
    monthly_revenue["monthly_trend"]
)

plt.figure(figsize=fig_size)
plt.plot(
    monthly_revenue["monthly_trend"],
    monthly_revenue["total_revenue"],
    marker="o",
    linewidth=2,
    color="#4682B4",
    markerfacecolor="#ffffff",
    markeredgewidth=2,
)

plt.draw()
plt.gca().get_xticklabels()[0].set_visible(False)

plt.grid(True, linestyle="--", alpha=0.3, color="#b0bec5")

plt.title("Monthly Revenue Trend", fontsize=14, color="#263238", pad=15)
plt.xlabel("Month", color="#37474f")
plt.ylabel("Revenue ($)", color="#37474f")
plt.tight_layout()
plt.savefig("images/monthly_revenue.png", dpi=150, bbox_inches="tight")
plt.show()
```

```python
top_customers = pd.read_csv("most_valuable_customers")

plt.figure(figsize=fig_size)
sns.barplot(
    data=top_customers,
    x="total_spent",
    y="name",
    hue="name",
    palette="mako",
    legend=False,
)

plt.title("Top 10 Customers by Spending")
plt.xlabel("Total Spent ($)")
plt.ylabel("")
plt.tight_layout()
plt.savefig("images/top_customers.png", dpi=150, bbox_inches="tight")
plt.show()
```

```python
top_rated = pd.read_csv("top_rated_products")

top_rated["name_with_rating"] = (
    top_rated["name"]
    + " (Rating: "
    + top_rated["average_rating"].astype(str)
    + ")"
)

plt.figure(figsize=fig_size)
sns.barplot(
    data=top_rated,
    x="total_reviews",
    y="name_with_rating",
    hue="name_with_rating",
    palette="magma",
    legend=False,
)

plt.title("Top Rated Products (Reviews & Ratings)")
plt.xlabel("Total Reviews")
plt.ylabel("")
plt.tight_layout()
plt.savefig("images/top_rated_products.png", dpi=150, bbox_inches="tight")
plt.show()
```

```python
country_sales = pd.read_csv("most_orders_country")

plt.figure(figsize=fig_size)
sns.barplot(
    data=country_sales,
    x="total_orders",
    y="country",
    hue="country",
    palette="flare",
    legend=False,
)

plt.title("Total Orders by Country")
plt.xlabel("Total Orders")
plt.ylabel("")
plt.tight_layout()
plt.savefig("images/country_sales.png", dpi=150, bbox_inches="tight")
plt.show()
```
